# Summary

```python
summary = pd.DataFrame({
    'start': df.iloc[0],
    'end': df.iloc[-1],
    'mean': df.mean(),
    'std': df.std()
})
summary['pct_change'] = (summary['end'] - summary['start']) / summary['start'] * 100
summary_sorted = summary.sort_values('pct_change', ascending = True)
print(summary_sorted)
```

## EDA Conclusions

**Overall dataset:**
- 51 instruments, 500 days of price history
- Position limits enforced by eval.py: $10,000 per instrument per, except ALGO allows $100,000 our biggest asset

**Instrument behavior (from summary stats):**
- Wide range of overall performance: from -60% (EAFC) to +114% (OTCS) over the 500 days
- ALGO sits near the middle (-13%), relatively lower volatility (~6.4% std relative to price)
- The flattest instrument based on pct_change is HTRK (~2.578%) with opening price 10.86 closed at11.14
- Some instruments trended consistently in one direction (e.g. OTCS up, EAFC down),
  others closer to range-bound

**Key observation from plotting:**
- Even the "trending" instruments (OTCS, EAFC) aren't smooth — they have real
  short-term swings/reversals within their longer-term direction
- This suggests a naive day-to-day momentum signal (buy whatever went up yesterday)
  may get whipsawed by short-term noise, even on instruments with a real underlying trend
- Possible directions to explore: longer lookback windows for momentum (smooth out noise),
  or blending momentum with mean-reversion elements depending on instrument behavior

Definition:
- Momentum strategy : betting that a recent price move continues (buy what went up, sell what went down). Works well on smooth trends, poorly on choppy/range-bound instruments.
- Mean-reversion strategy : betting that a price far from its average will return to it (buy when low, sell when high relative to its own mean). Works well on range-bound instruments, poorly on strong trenders.
- Lookback window : how far back a strategy looks when calculating a signal (e.g. "yesterday's return" is a 1-day lookback; "return over the past 20 days" is a 20-day lookback). Longer lookbacks smooth out short-term noise.
