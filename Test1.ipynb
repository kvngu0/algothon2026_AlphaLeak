```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# import prices.txt data
data_path = 'prices.txt'
df = pd.read_csv(data_path, sep=' ')
```

```python
print(f"Number of features: {len(df.columns)}")
print(f"Number of days: {df.shape[0]}")
```

```python
df
```

# EDA

```python
# Small plot
y1 = df['ALGO']
y2 = df['AENO']
x = np.arange(0, df.shape[0])

plt.figure(figsize=(6, 4))
plt.plot(x, y1, label="ALGO", linestyle="-")
plt.plot(x, y2, label="AENO", linestyle="-")
# for i in range(0, len(boundary_name)):
#     plt.plot(x, y_b[i], label=boundary_name[i], color=boundary_color.get(boundary_name[i]), linestyle="-")

plt.title(r"Asset Pricing")
plt.ylabel("Prices")
plt.xlabel("Days")
plt.legend(loc=0, fontsize='xx-small')
plt.show()
```

```python
# Large Plot
asset_list = df.columns

plt.figure(figsize=(10, 10))

for asset in asset_list:
    plt.plot(x, df[asset], label=asset, linestyle="-")

plt.title(r"Asset Pricing")
plt.ylabel("Prices")
plt.xlabel("Days")
plt.legend(loc=0, fontsize='xx-small')
plt.show()
```
